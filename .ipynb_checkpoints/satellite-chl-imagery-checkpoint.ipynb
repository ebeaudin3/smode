# Satellite Imagery of Surface Chlorophyll

```python
from smode_import import *

# load dopperscatt
ds_oce = xr.open_dataset('dopplerscatt_gridded.nc')
ds_wind = xr.open_dataset('dopplerscatt_wind_gridded.nc')
ds_ekman = xr.open_dataset('dopplerscatt_w_ekman_gridded.nc')

# load mesoscale currents
folder='/Users/elise/data/'
ncfile='cmems_obs-sl_glo_phy-ssh_my_allsat-l4-duacs-0.25deg_P1D_1715024829743.nc'
geos  = xr.open_dataset(folder+ncfile)

# sentinel 3
datadir='/Users/elise/data/SMODE-data/Sentinel3A/IOP2/'
sen3 = xr.open_dataset(datadir+'Sentinel_3A_20230422T184454Z.nc')
#sen3['CHL_OC4ME']        = sen3.CHL_OC4ME.where(sen3.CHL_OC4ME <= 100) #np.linspace(0.134,0.138,20)
sen3['CHL_OC4ME_real'] = 10**((sen3.CHL_OC4ME + 2) / 0.01574803)
sen3['CHL_OC4ME_real'] = sen3['CHL_OC4ME_real'].where(np.isfinite(sen3['CHL_OC4ME_real']))
sen3['lat'],sen3['lon']  = sen3.lat*1e6, sen3.lon*1e6

def gridlines(gl,delta=0.4):
    from cartopy.mpl.gridliner import LONGITUDE_FORMATTER, LATITUDE_FORMATTER
    import matplotlib.ticker as mticker
    
    gl.xlines = False
    gl.ylines = False
    
    gl.xlocator = mticker.FixedLocator(np.arange(-180,180,delta))
    gl.ylocator = mticker.FixedLocator(np.arange(-90,90,delta))

    gl.top_labels = False
    gl.right_labels = False
    
    gl.xformatter = LONGITUDE_FORMATTER
    gl.yformatter = LATITUDE_FORMATTER
    #gl.xlabel_style = {'size': 15, 'color': 'gray'}
    #gl.xlabel_style = {'color': 'red', 'weight': 'bold'}
```

```python
# Low resolution satellite chlorophyll (4km)
fontsize=14
plt.rcParams["font.size"] = fontsize
plt.rcParams["pdf.fonttype"] = 42

path='/Users/elise/data/SMODE-data/Satellite-Chlorophyll/'
data = xr.open_dataset(path+'20230422_cmems_obs-oc_glo_bgc-plankton_my_l4-gapfree-multi-4km_P1D.nc')
data = data.CHL.isel(time=0).sel(lat=slice(40, 20), lon=slice(-130, -120))

import matplotlib.colors as mcolors

fig,ax=plt.subplots(1,1,figsize=(8,8),subplot_kw={'projection': ccrs.PlateCarree()})

# chlorophyll concentration
vmin, vmax = 0.6,50 # 0.6, 50
norm = mcolors.LogNorm(vmin=vmin, vmax=vmax)
p = ax.contourf(data.lon, data.lat, data.values, np.linspace(vmin,vmax,200), cmap='Greens', norm=norm, extend='both')

# mesoscale field
geos_ = geos.sel(time='2023-04-22')
X, Y = np.meshgrid(geos_.longitude, geos_.latitude)
U, V = geos_.ugos.values, geos_.vgos.values
ax.streamplot(X, Y, U, V, color='k', density=8, linewidth=0.8, arrowsize=0.8)

# coastline and land
ax.coastlines(resolution='10m', color='black', linewidth=1.2, zorder=101)
ax.add_feature(cfeature.LAND, color='floralwhite',zorder=100)#color='PAPAYAWHIP')
ax.set_xlim(-126,-121); ax.set_ylim(35,39)
smode_pgon(plt.gca(),c='k',lw=2)
gl = ax.gridlines(crs=ccrs.PlateCarree(), draw_labels=True,
                  linewidth=2, color='gray', alpha=0.5, linestyle='--')

# other features
gridlines(gl, delta=1)
cbar=plt.colorbar(p,ax=ax,shrink=0.5)
cbar.set_ticks([1e0, 1e1])
cbar.set_ticklabels([r'$10^{0}$', r'$10^{1}$']) #logscale

for spine in ax.spines.values():
    spine.set_edgecolor('black')
    spine.set_linewidth(1.)
    spine.set_zorder(200)

#plt.savefig('/Users/elise/ocean/drifters_chl/high-res/fig.png',dpi=800, transparent=True)
#plt.savefig('/Users/elise/ocean/drifters_chl/high-res/fig1/fig1a.pdf', bbox_inches='tight')
plt.show()
```

```python
# Plot chl and dopplerscatt together
f = coriolis_parameter(35)

def smooth(data,sigma=5):
    import scipy.ndimage
    return scipy.ndimage.gaussian_filter(data,sigma)

fig,ax=plt.subplots(2,1,figsize=(5,10),subplot_kw={'projection': ccrs.PlateCarree()})

ax[0].contour(ds_oce.lon,ds_oce.lat,smooth(ds_oce.vorticity.sel(time=slice('2023-04-22','2023-04-23'))[0,...]),20,colors='k')
ax[1].contour(ds_oce.lon,ds_oce.lat,smooth(ds_oce.divergence.sel(time=slice('2023-04-22','2023-04-23'))[0,...]),10,colors='k')

# highlight the Ro=1 line
ax[0].contour(ds_oce.lon,ds_oce.lat,
              (smooth(ds_oce.vorticity.sel(time=slice('2023-04-22','2023-04-23'))[0,...]))/f,
              levels=[-2,-1,0,1,2],linewidths=2,colors='red')
ax[1].contour(ds_oce.lon,ds_oce.lat,
              (smooth(ds_oce.divergence.sel(time=slice('2023-04-22','2023-04-23'))[0,...]))/f,
              levels=[-2,-1,0,1,2],linewidths=2,colors='b')

#ax[0].contour(ds.lon,ds.lat,smooth(ds_wind.vorticity),20,colors='r')
#ax[1].contour(ds.lon,ds.lat,smooth(ds_wind.divergence),20,colors='r')

# add chlorophyll
for axis in ax:
    cf=axis.contourf(sen3.lon, sen3.lat, sen3['CHL_OC4ME_real'], np.linspace(vmin,vmax,200), cmap='Greens', norm=norm, extend='both')
    cf.set_rasterized(True)
    axis.set_xlim(-124.45,-123.9); axis.set_ylim(36.7,37.05)
    smode_pgon(axis,c='w')
    gl = axis.gridlines(crs=ccrs.PlateCarree(), draw_labels=True,
                  linewidth=2, color='gray', alpha=0.5, linestyle='--')
    gridlines(gl,delta=0.2)
    gl.xlabel_style = {'size': 12}; gl.ylabel_style = {'size': 12}; 
#plt.xlim(-125,-123.8); plt.ylim(36,39)
#plt.savefig('/Users/elise/Desktop/submesoscale.png',dpi=800, bbox_inches='tight', transparent=True)
plt.savefig('/Users/elise/ocean/drifters_chl/high-res/fig1/fig1bc_rasterized.pdf',transparent=True) 
plt.show()
```

## Eddy Transect

```python

time_crossing_intervals = [
    ['2023-04-20T03:20:00', '2023-04-20T05:00:00'],
    ['2023-04-20T21:00:00', '2023-04-20T23:00:00'],
    ['2023-04-21T16:40:00', '2023-04-21T17:25:00'],
    ['2023-04-21T18:30:00', '2023-04-21T19:26:00'],
    ['2023-04-22T21:55:00', '2023-04-22T23:40:00'],
    ['2023-04-24T21:40:00', '2023-04-24T23:30:00'],
    ['2023-04-24T21:40:00', '2023-04-24T23:30:00']
] # time we crossed the eddy

times_dopplerscatt=['2023-04-19T20','2023-04-20T20',
                  '2023-04-21T15','2023-04-21T21',
                  '2023-04-22T20','2023-04-24T20',]
orientations = ['zonal','meridional','meridional','zonal','meridional']
reverse=[False,False,False,False,True]
```

```python
from scipy.ndimage import gaussian_filter

# MVP & Dopplerscatt datasets
da = xr.open_dataset('MVP.nc')
da['rho'] = da.rho.where(da.salt >= 32.5) # keep density where salt < 32.5
ds = xr.open_dataset('dopplerscatt_gridded.nc') # dopplerscatt

# Select transect
i = 4
time_start = (datetime.strptime(time_crossing_intervals[i][0], '%Y-%m-%dT%H:%M:%S') - timedelta(minutes=0)).strftime('%Y-%m-%dT%H:%M:%S')
time_end = (datetime.strptime(time_crossing_intervals[i][1], '%Y-%m-%dT%H:%M:%S') + timedelta(minutes=0)).strftime('%Y-%m-%dT%H:%M:%S')

da_ = da.sel(time=slice(time_start, time_end))
lon0,lon1=da_.lon.min().values,da_.lon.max().values
lat0,lat1=da_.lat.min().values,da_.lat.max().values

# Panel 1: vorticity & divergence
# Plot
fig, ax = plt.subplots(4,1,figsize=(4,12),
                       gridspec_kw={'height_ratios': [0.5, 1, 1, 1,],'hspace': 0.15})
plt.rcParams.update({'font.size': 12})
axis=ax[0]

time_start = times_dopplerscatt[i]
time_end = (datetime.strptime(time_start, '%Y-%m-%dT%H') + timedelta(hours=10)).strftime('%Y-%m-%dT%H')
ds_ = ds.sel(time=slice(time_start, time_end)).isel(time=0) # div/vor
w_E = ds_ekman.sel(time=slice(time_start,time_end)).isel(time=0) # Ekman Transport
f = coriolis_parameter(37)

tick_dist_km = np.arange(0, 50, 5)
print(orientations[i])
if orientations[i]=='meridional':
    l1,=(ds_.sel(lon=lon0, method='nearest').divergence / f).plot(label=r'$\delta$', color='darkblue', linewidth=2, ax=axis)
    l2,=(ds_.sel(lon=lon0, method='nearest').vorticity / f).plot(label=r'$\zeta$', color='orangered', linewidth=2, ax=axis)
    ticks = lat0 + tick_dist_km / 111
elif orientations[i]=='zonal':
    l1,=(ds_.sel(lat=lat0, method='nearest').divergence / f).plot(label=r'$\delta/f$', color='darkblue', linewidth=2, ax=axis)
    l2,=(ds_.sel(lat=lat0, method='nearest').vorticity / f).plot(label=r'$\zeta/f$', color='orangered', linewidth=2, ax=axis)
    phi = np.mean(da_.lat.values)
    km_per_deg_lon = 111.32 * np.cos(np.deg2rad(phi))
    ticks = (lon_subset - lon0) * km_per_deg_lon

# x-ticks in distances [km]
axis.set_xticks(ticks)
axis.set_xticklabels(tick_dist_km)
axis.set_xlabel("Distance (km)")
if orientations[i]=='meridional': axis.set_xlim(lat0, lat1)
if orientations[i]=='zonal': axis.set_xlim(lon0, lon1)
axis.set_ylabel(r'($f^{-1}$)')
axis.spines['top'].set_color('w'); axis.spines['right'].set_color('w')
axis.legend(fontsize=10, loc='upper right', edgecolor='w')
axis.axhline(0, color='k', linestyle='--', alpha=0.5, linewidth=0.8)
if reverse[i]: axis.invert_xaxis()


# Transect panels: temp, salt, chl
n=40
p_temp=da.temp.plot.contourf(levels=np.linspace(9.5,12.,n),ax=ax[1],cmap=cm.thermal,x='time', add_colorbar=True)
p_salt=da.salt.plot.contourf(levels=np.linspace(32.8,33.8,n),ax=ax[2],cmap=cm.haline,x='time', add_colorbar=True)
p_chl =da.chla.plot.contourf(levels=np.linspace(0.5,7,n),ax=ax[3],cmap='Greens',x='time', add_colorbar=True)

# set cbar ticks
for cbar in [p_temp.colorbar, p_salt.colorbar, p_chl.colorbar]:
    cbar.ax.tick_params(length=0)
p_temp.colorbar.set_ticks(np.arange(9.5,12.5,0.5)); p_temp.colorbar.set_label(r'temp ($\degree $C)')
p_salt.colorbar.set_ticks(np.arange(32.8,33.9,0.2)); p_salt.colorbar.set_label('salt (psu)')
p_chl.colorbar.set_ticks(np.arange(0,8,2)); p_chl.colorbar.set_label(r'chl (mg m$^{-3}$)')
# add colorbar to ax0 for shape
plt.colorbar(p_temp,ax=ax[0])

# set x-limits
dt=0. #hours
for axis in ax[1::]:
    axis.set_xlim(pd.Timestamp(time_crossing_intervals[i][0])- timedelta(hours=dt),
                pd.Timestamp(time_crossing_intervals[i][1])+ timedelta(hours=dt))

# apply gaussian filter to density contours
rho_smoothed = da_.rho.rolling(time=3, center=True, min_periods=1).mean()
for axis in ax[1::]:
    #da_.rho.plot.contour(levels=20,x='time',colors='k',linewidths=0.8,ax=axis)
    rho_smoothed.plot.contour(levels=20,x='time',colors='k',linewidths=1.5,ax=axis)
    rho_smoothed.plot.contour(levels=[1025.3], x='time', colors='r',linewidths=2,ax=axis)
    axis.set_xlabel('')
    axis.invert_yaxis()
    axis.set_xticks(da_.time[::4], '')

# y-label
for axis in ax[1::]:
    axis.set_ylabel('depth (m)')

# remove titles
for axis in ax: axis.set_title('')

plt.savefig('/Users/elise/ocean/drifters_chl/high-res/fig1/fig1defg.pdf')#,transparent=True) 
plt.show()
```

```python
# Transect eddy-crossing

# MVP & Dopplerscatt datasets
da = xr.open_dataset('MVP.nc')
da['rho'] = da.rho.where(da.salt >= 32.5) # keep density where salt < 32.5
ds = xr.open_dataset('dopplerscatt_gridded.nc') # dopplerscatt

# Select transect
i = 4
da_ = da.sel(time=slice(time_crossing_intervals[i][0],time_crossing_intervals[i][1])) # Select crossing
lon0,lon1=da_.lon.min().values,da_.lon.max().values
lat0,lat1=da_.lat.min().values,da_.lat.max().values

# Plot
fig, ax = plt.subplots(5,1,figsize=(3,6*2), sharex=True)
plt.rcParams.update({'font.size': 12})

# Panel 1: vorticity & divergence
axis=ax[0]

time_start = times_dopplerscatt[i]
time_end = (datetime.strptime(time_start, '%Y-%m-%dT%H') + timedelta(hours=10)).strftime('%Y-%m-%dT%H')
ds_ = ds.sel(time=slice(time_start, time_end)).isel(time=0) # div/vor
w_E = ds_ekman.sel(time=slice(time_start,time_end)).isel(time=0) # Ekman Transport
f = coriolis_parameter(37)

tick_dist_km = np.arange(0, 25, 5)
if orientations[i]=='vertical':
    l1,=(ds_.sel(lon=lon0, method='nearest').divergence / f).plot(label=r'$\delta/f$', color='darkblue', linewidth=3, ax=axis)
    l2,=(ds_.sel(lon=lon0, method='nearest').vorticity / f).plot(label=r'$\zeta/f$', color='orangered', linewidth=3, ax=axis)
    axis.set_xlim(lat0, lat1)
elif orientations[i]=='horizontal':
    l1,=(ds_.sel(lat=lat0, method='nearest').divergence / f).plot(label=r'$\delta/f$', color='darkblue', linewidth=3, ax=axis)
    l2,=(ds_.sel(lat=lat0, method='nearest').vorticity / f).plot(label=r'$\zeta/f$', color='orangered', linewidth=3, ax=axis)
    axis.set_xlim(lon0, lon1)
# x-ticks in distances [km]

# plot features
if reverse[i]==1: axis.invert_xaxis()
axis.set_title('')
axis.set_xlabel('')
axis.set_ylim(-1,2)
axis.axhline(0, color='k')
# single legend
lines=[l1,l2]
axis.legend(lines, [line.get_label() for line in lines], 
            loc='upper right',framealpha=1,fontsize=10,)
axis.set_ylabel(r'$\zeta/f$, $\delta/f$')
if orientations[i]=='vertical': axis.set_xlim(lat0, lat1)
    
# Temp + Salt + Chl on other panel
n=40
p_temp=da.temp.plot.contourf(levels=np.linspace(9.5,12.,n),ax=ax[1],cmap=cm.thermal,x='time', add_colorbar=False)
p_salt=da.salt.plot.contourf(levels=np.linspace(32.8,33.8,n),ax=ax[2],cmap=cm.haline,x='time', add_colorbar=False)
p_chl =da.chla.plot.contourf(levels=np.linspace(0.5,7,n),ax=ax[3],cmap='Greens',x='time', add_colorbar=False)

dt=0. #hours
for axis in ax[1::]:
    axis.set_xlim(pd.Timestamp(time_crossing_intervals[i][0])- timedelta(hours=dt),
                pd.Timestamp(time_crossing_intervals[i][1])+ timedelta(hours=dt))
    
    da.rho.plot.contour(levels=20,x='time',colors='k',linewidths=0.8,ax=axis)
    da.rho.plot.contour(levels=[1025.3], x='time', colors='r',linewidths=2,ax=axis)
    axis.set_xlabel('')
    axis.invert_yaxis()
    axis.set_xticks(da_.time[::4], '')

plt.colorbar(p_temp,ax=ax[0])
cbar_temp=plt.colorbar(p_temp,ax=ax[1])
cbar_salt=plt.colorbar(p_salt,ax=ax[2])
cbar_chl=plt.colorbar(p_chl,ax=ax[3])

cbar_temp.set_ticks(np.round(cbar_temp.get_ticks()[::2],1))
cbar_salt.set_ticks(np.round(cbar_salt.get_ticks()[::2],1))
cbar_chl.set_ticks(np.round(cbar_chl.get_ticks()[::2],1))

plt.suptitle(times_dopplerscatt[i])

# Set x-limits for subsequent axes
for axis in ax[1::]:
    axis.set_yticks([0,25,50,75,100])
    axis.set_xticks(da_.time[::4])  # Set tick positions
    if orientations[i]=='vertical':
        axis.set_xticklabels(np.round(da_.lat[::4].values,2))  # Set the labels
    if orientations[i]=='horizontal':
        axis.set_xticklabels(np.round(da_.lon[::4].values,2))  # Set the labels
    
    axis.set_xlim(da_.time[0], da_.time[-1])
    axis.set_ylabel('depth (m)',fontsize=12)

#for axis in ax[0:-1]:
#    axis.set_xticks([])

#plt.savefig('/Users/elise/Desktop/fig3defg.png',dpi=800, bbox_inches='tight', transparent=True)
plt.tight_layout()
plt.show();
```

```python
def gridlines(gl,delta=0.4):
    from cartopy.mpl.gridliner import LONGITUDE_FORMATTER, LATITUDE_FORMATTER
    import matplotlib.ticker as mticker
    
    gl.xlines = False
    gl.ylines = False
    
    gl.xlocator = mticker.FixedLocator(np.arange(-180,180,delta))
    gl.ylocator = mticker.FixedLocator(np.arange(-90,90,delta))

    gl.top_labels = False
    gl.right_labels = False
    
    gl.xformatter = LONGITUDE_FORMATTER
    gl.yformatter = LATITUDE_FORMATTER
    #gl.xlabel_style = {'size': 15, 'color': 'gray'}
    #gl.xlabel_style = {'color': 'red', 'weight': 'bold'}
```
