```python
from smode_import import *
import cartopy.crs as ccrs

# load dopperscatt
ds_oce = xr.open_dataset('../dopplerscatt_gridded.nc')

# sentinel 3
datadir='/Users/elise/data/SMODE-data/Sentinel3A/IOP2/'
sen3 = xr.open_dataset(datadir+'Sentinel_3A_20230422T184454Z.nc')
sen3['CHL_OC4ME_real'] = 10**((sen3.CHL_OC4ME + 2) / 0.01574803)
sen3['CHL_OC4ME_real'] = sen3['CHL_OC4ME_real'].where(np.isfinite(sen3['CHL_OC4ME_real']))
sen3['lat'],sen3['lon']  = sen3.lat*1e6, sen3.lon*1e6

def gridlines(gl,delta=0.4):
    from cartopy.mpl.gridliner import LONGITUDE_FORMATTER, LATITUDE_FORMATTER
    import matplotlib.ticker as mticker
    
    gl.xlines = False
    gl.ylines = False
    
    gl.xlocator = mticker.FixedLocator(np.arange(-180,180,delta))
    gl.ylocator = mticker.FixedLocator(np.arange(-90,90,delta))

    gl.top_labels = False
    gl.right_labels = False
    
    gl.xformatter = LONGITUDE_FORMATTER
    gl.yformatter = LATITUDE_FORMATTER
    #gl.xlabel_style = {'size': 15, 'color': 'gray'}
    #gl.xlabel_style = {'color': 'red', 'weight': 'bold'}
```

```python
import matplotlib.colors as mcolors
vmin, vmax = 0.6,50
norm = mcolors.LogNorm(vmin=vmin, vmax=vmax)
```

```python
# Plot chl and dopplerscatt together
f = coriolis_parameter(35)

def smooth(data,sigma=5):
    import scipy.ndimage
    return scipy.ndimage.gaussian_filter(data,sigma)

fig,ax=plt.subplots(2,1,figsize=(3.85,4.92),subplot_kw={'projection': ccrs.PlateCarree()})
fig.subplots_adjust(left=0.17, right=0.985, bottom=0.06, top=0.985, hspace=0.28)

#ax[0].contour(ds_oce.lon,ds_oce.lat,smooth(ds_oce.vorticity.sel(time=slice('2023-04-22','2023-04-23'))[0,...]),20,colors='k')
#ax[1].contour(ds_oce.lon,ds_oce.lat,smooth(ds_oce.divergence.sel(time=slice('2023-04-22','2023-04-23'))[0,...]),10,colors='k')

# Fifth-degree lines carry the eddy structure. Integers are colored;
# integers and half-integers are numbered.
all_levels = np.round(np.arange(-2, 2.01, 0.2), 2)

def level_kind(lev):
    if abs(lev - round(lev)) < 1e-6:
        return 'int'
    if abs(2*lev - round(2*lev)) < 1e-6:
        return 'half'
    return 'step'

structure = [lv for lv in all_levels if level_kind(lv) == 'step']
named = [lv for lv in all_levels if level_kind(lv) != 'step']
vort_norm = smooth(ds_oce.vorticity.sel(time=slice('2023-04-22','2023-04-23'))[0,...]) / f
div_norm  = smooth(ds_oce.divergence.sel(time=slice('2023-04-22','2023-04-23'))[0,...]) / f
xlim, ylim = (-124.45, -123.9), (36.7, 37.05)

def named_style(color):
    colors, lws = [], []
    for lev in named:
        if level_kind(lev) == 'int':
            colors.append(color); lws.append(2)
        else:
            colors.append('k'); lws.append(2)
    return colors, lws

def cfmt(v):
    if abs(v) < 1e-8:
        return '0'
    s = f'{v:+.0f}' if abs(abs(v) - round(abs(v))) < 1e-8 else f'{v:+.1f}'
    return s.replace('-', '\u2212')

def label_spots(cs):
    # one upright label on each long piece, kept off the frame and off each other
    side = {-0.5: 'north', 0: 'south', 0.5: 'south', 1: 'west', 1.5: 'south', 2: 'north'}
    nmax = {0: 1, -0.5: 2, 0.5: 2, 1: 3, 1.5: 1, 2: 1}
    spots = []
    for lev, segs in zip(cs.levels, cs.allsegs):
        which = side.get(round(float(lev), 1), 'south')
        cands = []
        for seg in segs:
            x, y = seg[:, 0], seg[:, 1]
            inside = ((x > xlim[0]+0.04) & (x < xlim[1]-0.04) &
                      (y > ylim[0]+0.025) & (y < ylim[1]-0.03))
            if inside.sum() < 8:
                continue
            L = np.nansum(np.hypot(np.diff(x)*88.5, np.diff(y)*111))
            if L < 6:
                continue
            ii = np.where(inside)[0]
            if which == 'south': j = ii[np.argmin(y[ii])]
            elif which == 'north': j = ii[np.argmax(y[ii])]
            elif which == 'west': j = ii[np.argmin(x[ii])]
            else: j = ii[np.argmax(x[ii])]
            cands.append((L, x[j], y[j]))
        cands.sort(reverse=True)
        kept = []
        for L, x, y in cands:
            if all(np.hypot((x-px)*88, (y-py)*111) > 12 for px, py in kept):
                kept.append((x, y))
            if len(kept) == nmax.get(round(float(lev), 1), 1):
                break
        spots.extend(kept)
    return spots

def draw_field(axis, field, color):
    axis.contour(ds_oce.lon, ds_oce.lat, field, levels=structure,
                 colors='k', linewidths=1.8, zorder=3)
    colors, lws = named_style(color)
    print(lws)
    cs = axis.contour(ds_oce.lon, ds_oce.lat, field, levels=named,
                      colors=colors, linewidths=lws, zorder=4)
    spots = label_spots(cs)
    write_numbers=0
    if len(spots) and write_numbers:
        texts = axis.clabel(cs, inline=True, inline_spacing=2, fontsize=8,
                            fmt=cfmt, colors='k', manual=spots, rightside_up=True)
        import matplotlib.patheffects as pe
        for t in texts:
            t.set_path_effects([pe.withStroke(linewidth=3, foreground='white')])
            t.set_zorder(6)
            t.set_clip_on(True)
    axis.set_aspect('auto')

# add chlorophyll, then the contours so the labels sit on the finished map
for axis in ax:
    cf=axis.contourf(sen3.lon, sen3.lat, sen3['CHL_OC4ME_real'], np.linspace(vmin,vmax,200), cmap='Greens', norm=norm, extend='both')
    cf.set_rasterized(True)
    axis.set_xlim(*xlim); axis.set_ylim(*ylim)
    axis.set_aspect('auto')  # panel frame matches fig1.pdf B/C (wider than a geographic aspect)
    smode_pgon(axis,c='w')
    gl = axis.gridlines(crs=ccrs.PlateCarree(), draw_labels=True,
                  linewidth=2, color='gray', alpha=0.5, linestyle='--')
    gridlines(gl,delta=0.2)
    gl.xlabel_style = {'size': 12}; gl.ylabel_style = {'size': 12};
    # tick marks on the frame, pointing out toward the lon/lat labels
    axis.set_xticks([-124.4, -124.2, -124.0], crs=ccrs.PlateCarree())
    axis.set_yticks([36.8, 37.0], crs=ccrs.PlateCarree())
    axis.tick_params(axis='both', which='major', direction='out', length=4, width=0.8,
                     bottom=True, left=True, top=False, right=False,
                     labelbottom=False, labelleft=False, colors='k')

# Panel B vorticity in red, panel C divergence in blue. Color marks integers only.
draw_field(ax[0], vort_norm, 'red')
draw_field(ax[1], div_norm, 'blue')
#plt.xlim(-125,-123.8); plt.ylim(36,39)
#plt.savefig('/Users/elise/Desktop/submesoscale.png',dpi=800, bbox_inches='tight', transparent=True)
#plt.savefig('/Users/elise/ocean/drifters_chl/high-res/fig1/fig1bc_rasterized.pdf',transparent=True) 
plt.savefig('/Users/elise/Desktop/SMODE-manuscript/figures/fig1bc.png',dpi=800, bbox_inches='tight', transparent=True)
plt.show()
```

```python
## import numpy as np
import scipy.ndimage
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib.lines import Line2D
import cartopy.crs as ccrs

# --- Setup Parameters & Smoothing ---
f = coriolis_parameter(35)

def smooth(data, sigma=5):
    return scipy.ndimage.gaussian_filter(data, sigma)

# Extract and smooth normalized vorticity and divergence fields
time_slice = slice('2023-04-22', '2023-04-23')
vort_norm = smooth(ds_oce.vorticity.sel(time=time_slice)[0, ...]) / f
div_norm  = smooth(ds_oce.divergence.sel(time=time_slice)[0, ...]) / f

# Define consistent background contour levels across both fields
contour_levels = np.arange(-2.0, 2.25, 0.25)

# Setup Figure
fig, ax = plt.subplots(2, 1, figsize=(5, 10), subplot_kw={'projection': ccrs.PlateCarree()})

# ==========================================
# PANEL B: VORTICITY (\zeta / f)
# ==========================================
# 1. Background Chlorophyll Field
cf0 = ax[0].contourf(sen3.lon, sen3.lat, sen3['CHL_OC4ME_real'], 
                     np.linspace(vmin, vmax, 200), cmap='Greens', norm=norm, extend='both')
cf0.set_rasterized(True)

# 2. Fine Background Contour Structure (thin black lines)
ax[0].contour(ds_oce.lon, ds_oce.lat, vort_norm, 
               levels=contour_levels, colors='k', linewidths=0.5, alpha=0.5)

# 3. Highlight Key Physical Thresholds
ax[0].contour(ds_oce.lon, ds_oce.lat, vort_norm, 
               levels=[0], colors='red', linewidths=1.5, linestyles='-')
ax[0].contour(ds_oce.lon, ds_oce.lat, vort_norm, 
               levels=[1], colors='red', linewidths=2.5, linestyles='-')

# 4. Clean In-Plot Legend Key
leg_b = [
    Line2D([0], [0], color='red', lw=1.5, ls='-', label=r'$\zeta/f = 0$'),
    Line2D([0], [0], color='red', lw=2.5, ls='-', label=r'$\zeta/f = +1$')
]
ax[0].legend(handles=leg_b, loc='upper right', fontsize=9, framealpha=0.9, edgecolor='none')


# ==========================================
# PANEL C: DIVERGENCE (\delta / f)
# ==========================================
# 1. Background Chlorophyll Field
cf1 = ax[1].contourf(sen3.lon, sen3.lat, sen3['CHL_OC4ME_real'], 
                     np.linspace(vmin, vmax, 200), cmap='Greens', norm=norm, extend='both')
cf1.set_rasterized(True)

# 2. Fine Background Contour Structure (thin black lines)
ax[1].contour(ds_oce.lon, ds_oce.lat, div_norm, 
               levels=contour_levels, colors='k', linewidths=0.5, alpha=0.5)

# 3. Highlight Key Physical Thresholds
ax[1].contour(ds_oce.lon, ds_oce.lat, div_norm, 
               levels=[0], colors='blue', linewidths=1.5, linestyles='-')
ax[1].contour(ds_oce.lon, ds_oce.lat, div_norm, 
               levels=[-0.5], colors='blue', linewidths=2.5, linestyles='--')

# 4. Clean In-Plot Legend Key
leg_c = [
    Line2D([0], [0], color='blue', lw=1.5, ls='-', label=r'$\delta/f = 0$'),
    Line2D([0], [0], color='blue', lw=2.5, ls='--', label=r'$\delta/f = -0.5$')
]
ax[1].legend(handles=leg_c, loc='upper right', fontsize=9, framealpha=0.9, edgecolor='none')


# ==========================================
# FORMATTING & MAP EXTENTS
# ==========================================
for axis in ax:
    axis.set_xlim(-124.45, -123.9)
    axis.set_ylim(36.7, 37.05)
    smode_pgon(axis, c='w')
    
    gl = axis.gridlines(crs=ccrs.PlateCarree(), draw_labels=True,
                        linewidth=1.5, color='gray', alpha=0.5, linestyle='--')
    gridlines(gl, delta=0.2)
    gl.xlabel_style = {'size': 12}
    gl.ylabel_style = {'size': 12}

plt.savefig('../fig1bc_rasterized.pdf', 
            transparent=True, bbox_inches='tight')
plt.show()
```
