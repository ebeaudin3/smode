```python
# Plot gliders with color = velocity
# compute glider velocity
# v(x,y)=dx(x,y)/dt  -> we have dx(x,y) and dt

def drift_velocity(drift):
    keep=drift.position_QCflag==1 #keep good data
    time,lon,lat=drift.time[keep],drift.longitude[keep],drift.latitude[keep]
    
    dlon = np.diff(lon)
    dlat = np.diff(lat)
    
    dx = dlon * np.cos(np.deg2rad(lat[1::].values)) * 111320. # convert to meters
    dy = dlat * 111320.
    
    dt = np.diff(time) / np.timedelta64(1, 's') # seconds
    
    u = dx/dt
    v = dy/dt
    
    U = np.sqrt(u**2 + v**2)
    
    return time[0:-1],lat[0:-1],lon[0:-1],u,v,U
```

```python
def save_drift_velocity(time, lat, lon, u, v, drift):
    data_dict = {
        'u': xr.DataArray(u, dims=('time'), coords={'time': time}, attrs={'long_name': 'zonal velocity', 'units': 'm/s'}),
        'v': xr.DataArray(v, dims=('time'), coords={'time': time}, attrs={'long_name': 'meridional velocity', 'units': 'm/s'}),
        'lat': xr.DataArray(lat, dims=('time'), coords={'time': time}, attrs={'long_name': 'latitude', 'units': 'degrees'}),
        'lon': xr.DataArray(lon, dims=('time'), coords={'time': time}, attrs={'long_name': 'longitude', 'units': 'degrees'})
    }
    dataset = xr.Dataset(data_dict)
    dataset.attrs = drift.attrs
    ncfile=f'~/ocean/drifters_chl/drift_vel/data/drift_vel_{drift.drifter_id}.nc'
    #print(ncfile)
    dataset.to_netcdf(ncfile)
    
    return 
```

```python
for file in files:
    drift=xr.open_dataset(file) # load file
    time,lat,lon,u,v,U = drift_velocity(drift) # compute velocities
    save_drift_velocity(time, lat, lon, u, v, drift) # save netcdf
```

```python
plt.figure()
plt.plot(lon[0],lat[0],'k*',markersize=12)
plt.scatter(lon[0:-1],lat[0:-1],c=U,cmap='jet',s=3,vmin=0,vmax=1)
cbar=plt.colorbar()
cbar.set_label('velocity (ms-1)')
plt.title(f'drifter {drift.drifter_id}');
#plt.savefig(f'drift_vel/{drift.drifter_id}.png',dpi=800);
```

# Power spectral density
for each drifter:
1. interpolate u,v in time on a regular sampling frequency ($f_s$)
2. fourier transform and plot Power vs frequency
3. high-pass, low-pass and band-pass filter velocities.. what are the cut-off frequencies?
4. compute Kinetic Energy \
$ MKE = 1/2 (u_{mean}^2 + v_{mean}^2)$ \
$ LKE = 1/2 (u_{lp}^2 + v_{lp}^2)$ \
$ EKE = 1/2 (u_{bp}^2 + v_{bp}^2)$ \
$ HKE = 1/2 (u_{hp}^2 + v_{hp}^2)$


Near-inertial wave (NIW): NIWs appear nearly everywhere in the ocean as a spectral peak at and just above the local inertial period f.
